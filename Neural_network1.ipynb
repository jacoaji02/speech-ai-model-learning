```python
"""Imagine you work as a Risk Analyst at a major bank.
Your team needs to build an automated system that looks at a credit card transaction and instantly flags the probability that it is fraudulent.
You decide to look at three specific input features for any incoming transaction: Distance from home, How many miles away from the cardholder's home address is this purchase happening?
Transaction Size: Is the amount significantly higher than their usual average purchase?
Time of Day: Is the transaction happening during unusual hours (like 3:00 AM)?
How the Layer Processes This:Your network has a hidden layer with two distinct specialized neurons.
You train them to look out for different patterns:
Neuron A (The "Stolen Card" Specialist): It places a high positive weight on Distance and Transaction Size. If someone is buying a $4,000 TV 600 miles away, this neuron's weighted sum spikes.
Neuron B (The "Account Takeover" Specialist): It cares deeply about Time of Day combined with Transaction Size. If a high-value transaction happens in the middle of the night, its weighted sum fires up.
Each neuron also has a bias ($b$) to make sure it doesn't overreact to every single minor variance (like a midnight snack run down the street).Finally, the raw scores are fed into a Sigmoid Activation Function.
This squashes the numbers into a clean percentage between $0$ and $1$ (e.g., $0.94$), representing a 94% probability that this transaction is fraud. Your bank can now use this output to instantly freeze the card!"""
```

```python
import numpy as np

# 1. Inputs (X): The transaction details
# [Distance from home = 45.0 miles, Size relative to normal = 3.5x, Hour of day = 3.0 AM]
X = np.array([45.0, 3.5, 3.0])

# 2. Weights (W): Matrix shape (3 inputs, 2 hidden neurons)
W = np.array([
    [0.12,  0.02],  # Weights for Distance
    [0.85,  0.95],  # Weights for Transaction Size
    [-0.10, 0.70]   # Weights for Time of Day
])

# 3. Biases (b): One scalar bias for each of the 2 hidden neurons
b = np.array([-0.5, -1.2])
```

```python
""" 1) Calculate the Weighted Sum Matrix: Multiply the input vector X by the weight matrix W and add the bias vector b. (Hint: Remember your matrix dot product tool np.dot()).
2) Create a Sigmoid Function: Write a custom reusable function named sigmoid(z) that calculates $\frac{1}{1 + e^{-z}}$ using np.exp().
Generate the Final Activated Output ($A$): Pass your calculated $Z$ matrix through your sigmoid function to get the final probabilities for both fraud-detection neurons.
Print your results to see the final probability scores."""
```

```python
multi_input_weight = np.dot(X, W)
Z = multi_input_weight + b
print("weighted_sum_matrix: ", Z)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print(sigmoid(Z))
```

```python
def tanh(z):
  return np.tanh(z)

def relu(z):
  return np.maximum(0, z)
D = np.array([-2, -1, 0, 1, 2])

sigmoid_function = sigmoid(D)
tanh_function = tanh(D)
relu_function = relu(D)

print("Sigmoid function: ", sigmoid_function)
print("tanh Function", tanh_function)
print("ReLU Function", relu_function)
```
