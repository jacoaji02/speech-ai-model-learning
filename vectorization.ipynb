```python
import numpy as np
import torch

sentences = [
    "I love machine learning",
    "NLP is amazing and fun",
    "I love programming Python",
    "Deep learning is deep",
    "Python is fun for NLP"
]

# --- Exercise 1: Tokenize ---
# TODO: Split the sentences above into lists of lowercased words.
tokenized_sentences = [ s.lower().split() for s in sentences ]
vocabulary = []

for sentence in tokenized_sentences:
    for word in sentence:
        vocabulary.append(word)


unique_words = sorted(list(set(vocabulary)))

# --- Exercise 2: Build a Vocabulary Map ---
# TODO: Extract every unique word from your tokens and map them to a unique ID number.
# Hint: Start with "[PAD]": 0 to leave room for padding sequences later.
vocab = {"[PAD]": 0}

for index, word in enumerate(unique_words):
    vocab[word] = index + 1



# --- Exercise 3: Convert Text to Token IDs ---
# TODO: Substitute the words in your sentences with their matching numeric index from your vocab map.
token_ids_dataset = []

for sentence in tokenized_sentences:
    token_ids = []
    for word in sentence:
        token_ids.append(vocab[word])
    token_ids_dataset.append(token_ids)

print(token_ids_dataset)


# --- Exercise 4 & 5: Random Embeddings & Vector Similarity ---
# TODO: Create a random NumPy array matrix of shape (vocab_size, embedding_dimension=4).
# Then, slice out the vector for two words and calculate their Cosine Similarity.
np.random.seed(42)

embedding_dimension = 4
vocab_size = len(vocab)

embedding_matrix = np.random.rand(vocab_size, embedding_dimension)
print(embedding_matrix)

def cosine_similarity(v1, v2):
    # TODO: Implement the cosine similarity math equation: (A • B) / (||A|| * ||B||)
    dot_product = np.dot(v1,v2)

    normalized_v1 = np.linalg.norm(v1)
    normalized_v2 = np.linalg.norm(v2)

    similarity = dot_product / (normalized_v1 * normalized_v2)

    return similarity

print(cosine_similarity(embedding_matrix[1], embedding_matrix[2]))
```
