```python
import pandas as pd
import numpy as np

data = {
    'Name' : ['Ajin', 'Abin', 'Alan', 'Amal'],
    'Age' : [20, 20, 18, 23],
    'City' : ['chennai', 'mumbai', 'kochi', 'blr'],
    'Marks' : [85, 90, 75, 56]
}

df = pd.DataFrame(data)
print(df)
```

```python
filter_3 = df[(df['Age'] > 20) & (df['Marks'] > 70)]['Name']
print(filter_3)
```

```python
conditions = [
    (df['Marks'] >= 90),
    (df['Marks'] >= 70) & (df['Marks'] < 90),
    (df['Marks'] >= 60) & (df['Marks'] < 70),
    (df['Marks'] < 60)
]

values = ['A', 'B', 'C', 'D']

df['Grade'] = np.select(conditions, values, default='F')
print(df)
```

```python
df['Status'] = np.where(df['Marks'] >= 60, 'Pass', 'Fail')
print(df)
```
