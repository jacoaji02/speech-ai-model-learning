```python
"""Exercise 1

Create variables to store a student’s:

name
age
city

Then print a proper introduction sentence using these values."""

name = 'ajin'
age = 23
city = 'rajapuram'
print(f'The student name is {name}, he is {age} old and he is from {city}')
```

```python
"""Exercise 2

Now extend your program:
Assume the student has marks in 3 subjects.

Store marks in three variables
Calculate and print total marks and average marks
Exercise 3

Modify your program:
Take student’s name and favorite subject as input from the user.
Print a personalized message using this information."""
subjects = {
"malayalam": 99,
"english" : 95,
"maths" : 98
}

total = subjects["malayalam"] + subjects['english'] + subjects['maths']

# average = total / 3
# print(total)
# print(average)

# student_name = input('enter student name: ')
# fav_subject = input('enter subject name: ')
# if fav_subject in subjects:
#     print(f'{student_name} got {subjects[fav_subject]} marks in {fav_subject}')
```

```python
"""🔵 Phase 2: Multiple Students (Lists + Loops Start)
Exercise 4

Now instead of one student:

Create a list of 5 student names
Print each student name using a loop"""

students = ['ajin', 'abin', 'allu', 'anu', 'nanu']
for student in students:
  print(student)
```

```python
"""Exercise 5

For the same 5 students:

Assume each student has a score list (5 students → 1 score each for now)
Store their marks in a list
Print each student with their score using a loop"""

marks = [50, 85, 98, 90, 96]
for student, mark in zip(students, marks):
  print(f'{student} : {mark}')

```

```python
"""Exercise 6

Find:

The total sum of all student scores
The average score of the class"""
total = sum(marks)
num_students = len(students)
average = total / num_students
print(f"total marks = {total}")
print(f'average mark = {average}')
```

```python
"""Exercise 7

Identify:

The highest scoring student (from the list)"""
highest = 0
for mark in marks:
  if mark >= highest:
    highest = mark
print(highest)
for student, mark in zip(students, marks):
  if highest == mark:
    print(f'{student} got highest mark ({mark})')
```

```python
"""Exercise 8

Create a function:

calculate_average(marks)
It takes a list of marks and returns the average

Use it to calculate class average."""

def calculate_average(marks):
  average = total / num_students
  return average

calculate_average(marks)
```

```python
"""Exercise 9

Create a function:

find_topper(names, marks)
It returns the name of the student with highest marks

Use it with your existing lists.
"""
def find_topper(students, marks):
  highest = 0
  for mark in marks:
    if mark >= highest:
      highest = mark
  for student, mark in zip(students, marks):
    if highest == mark:
      return f'{student} got highest mark {mark}'

find_topper(students, marks)
```

```python
"""Create a function:

is_pass(mark)
Returns True if mark ≥ 40 else False

Use it to check each student’s result.
"""

def is_pass(students, marks):
  pass_mark = 60
  for student, mark in zip(students, marks):
    if mark >= pass_mark:
      print(f'{student} passed')
    else:
      print(f'{student} failed')


is_pass(students, marks)




```

```python
"""Exercise 13

For each student:

Add a new key "result" → "Pass" or "Fail" using is_pass() function"""

ages = [23, 24, 25, 26, 27]

student_data_list = [] # Initialize an empty list to store student dictionaries

for student, age, mark in zip(students, ages, marks):
  student_data = {
      'name': student,
      'age': age,
      'score': mark
  }
  student_data_list.append(student_data) # Append each student's dictionary to the list

print(student_data_list)
```

```python
"""Exercise 14

Update each student dictionary by adding:

"grade" based on marks:
90+ → A
75–89 → B
50–74 → C
below 50 → D"""

for item in student_data_list:
  if item['score'] >= 90:
    item['grade'] = 'A'
  elif item['score'] >= 75:
    item['grade'] = 'B'
  elif item['score'] <= 50:
    item['grade'] = 'C'
  else:
    item['grade'] = 'D'

print(student_data_list)
```

```python
"""Exercise 15

Count:

How many students passed
How many failed"""

pass_count = 0
fail_count = 0

for item in student_data_list:
  if item['score'] > 60:
    pass_count = pass_count + 1
  else:
    fail_count = fail_count + 1

print(f'pass count = {pass_count}')
print(f'fail count = {fail_count}')



```

```python
"""Exercise 16

Find:

Highest scorer
Lowest scorer"""

highest_mark = 0
for item in student_data_list:
  if item['score'] >= highest_mark:
    highest_mark = item['score']
for item in student_data_list:
  if item['score'] == highest_mark:
    print(f'{item["name"]} got highest mark {highest_mark}')

lowest_mark = 100
for item in student_data_list:
  if item['score'] <= lowest_mark:
    lowest_mark = item['score']
for item in student_data_list:
  if item['score'] == lowest_mark:
    print(f'{item["name"]} got lowest mark {lowest_mark}')

```

```python
"""Print a report:
For each student show:

Name
Marks
Grade
Result"""

print(student_data_list)

for item in student_data_list:
  if item['grade'] == 'A' or item['grade'] == 'B':
    item['result'] = 'Pass'
  else:
    item['result'] = 'Fail'

print(student_data_list)

for item in student_data_list:
  print(f'Name: {item['name']}')
  print(f'Mark: {item['score']}')
  print(f'Grade: {item['grade']}')
  print(f'Result: {item['result']}\n')
```
