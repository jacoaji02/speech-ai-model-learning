```python
from transformers import AutoTokenizer
import time

# 1. Load the fast, Rust-backed tokenizer
print("Loading Fast Tokenizer...")
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased", use_fast=True)
print(class_name := type(tokenizer).__name__)
# assert "Fast" in class_name, "Ensure you are using the Rust-backed Fast version!"

# 2. Define our real-time examination strings
test_cases = {
    "Standard English": "Transformers are changing the world of AI.",
    "BPE Subword Splitting": "That megastructure looks completely unbelievable.",
    "Malayalam Morphology": "അവർ സ്കൂളിൽനിന്നും വീട്ടിലേക്ക് നടന്നു."
}

print("\n--- Running Real-Time Examination ---")
for category, text in test_cases.items():
    print(f"\nCategory: {category}")
    print(f"Raw Input: '{text}'")

    # Track execution speed to observe Rust backend efficiency
    start_time = time.perf_counter()
    tokens = tokenizer.tokenize(text)
    input_ids = tokenizer.encode(text)
    end_time = time.perf_counter()

    print(f"Subword Tokens : {tokens}")
    print(f"Numerical IDs  : {input_ids}")
    print(f"Execution Time : {(end_time - start_time) * 1000:.4f} ms")
    print("-" * 40)
```

```python
from transformers import AutoTokenizer

# 1. Load the Multilingual BERT Tokenizer (Contains Malayalam in its vocabulary pool)
multilingual_tokenizer = AutoTokenizer.from_pretrained("l3cube-pune/malayalam-bert")

malayalam_text = "കമ്പ്യൂട്ടറുകളിൽനിന്നും"

# 2. Tokenize the text
tokens = multilingual_tokenizer.tokenize(malayalam_text)
token_ids = multilingual_tokenizer.encode(malayalam_text)

print(f"✨ Multilingual Vocab Size: {multilingual_tokenizer.vocab_size}")
print(f"🧩 Subword Tokens : {tokens}")
print(f"🔢 Numerical IDs  : {token_ids}")
```
